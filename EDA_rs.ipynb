Com os dados consistêntes, realizamos a agregação dos dados, que é a soma das principais métricas.
Por que essas e não outras variáveis? Por que essas são compostas pelo valor de outras variáveis.
A agregação possíbilita vermos total da carteira de crédito do RS, no mês escolhido.

```python
metricas = [
    'carteira_ativa',
    'carteira_a_vencer',
    'carteira_vencida',
    'carteira_inadimplencia',
    'ativo_problematico'
]

rs_agregado = df_rs[metricas].sum()
rs_agregado
```

Inicialmente, a análise é conduzida no nível agregado do estado, com o objetivo de estabelecer um baseline da carteira de crédito. Esse referencial permite, em etapas posteriores, avaliar desvios e assimetrias ao segmentar a carteira por perfil do tomador ou tipo de operação.

```python
pct_carteira_vencida = rs_agregado["carteira_vencida"] / rs_agregado["carteira_ativa"]
pct_carteira_vencida_RS = round(float(pct_carteira_vencida * 100), 2)
pct_carteira_vencida_RS
```

O percentual de carteira vencida é um indicador inicial de deterioração da carteira, refletindo atrasos no pagamento. Isoladamente, não caracteriza inadimplência, mas funciona como sinal de alerta quando analisado em conjunto com métricas de inadimplência e ativos problemáticos.
Como vimos o 1.72% da carteira ativa está vencida, um risco baixo para carteira.

```python
pct_carteira_inadimplente = rs_agregado["carteira_inadimplencia"] / rs_agregado["carteira_ativa"]
pct_carteira_inadimplente_RS = round(float(pct_carteira_inadimplente * 100), 2)
pct_carteira_inadimplente_RS
```

A carteira ativa do RS tem um percentual de inadimplência de 2.35%. Ou seja, a cada R$100 de crédito, R$2,35 não são pagos. Nesse primeiro momento não temos como dizer se esse valor é alto ou não, ou se está dentro do esperado, pois não temos um valor base para fazer essa comparação. O que ele mostra até aqui é que há inadimplência que não pode ser deixada de lado.

```python
pct_carteira_ativo_problemático = rs_agregado["ativo_problematico"] / rs_agregado["carteira_ativa"]
pct_carteira_ativo_problemático_rs = round(float(pct_carteira_ativo_problemático * 100), 2)
pct_carteira_ativo_problemático_rs
```

A carteira ativa possui 5.79% de ativos problemáticos para o mês. Que são aqueles que possúem algum atraso acima de 90 dias ou que passaram por alguma deteriorização da qualidade de crédito.
Ao olhar isoladamente para eles, apresentam um risco não concretizado, é mais uma ameaça a carteira ativa. Quando olharmos em comparação com outros meses será possível estabeler uma relação dele com outros indicadores.
